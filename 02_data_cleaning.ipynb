# EV Charging Dataset — Data Cleaning

## Objective

The objective of this stage is to identify and handle data quality issues in the EV Charging Dataset.

The cleaning process will include:
1. Checking duplicate records.
2. Investigating negative and zero charging durations.
3. Validating charging timestamps.
4. Converting date and time columns into appropriate data types.
5. Checking invalid or inconsistent values.
6. Saving a cleaned dataset for Exploratory Data Analysis (EDA).

The original CSV file will remain unchanged.

# 1.Load the dataset

```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("Data_row/ChargingRecords.csv")
df.head()
```

```python
df.shape
```

# 2. Check Duplicated Record

```python
# count duplicate rows
duplicate_count =df.duplicated().sum()

print("Total duplicate records", duplicate_count)
```

# Remove Duplicate Records



```python
# Remove duplicate rows
df = df.drop_duplicates()

# check remaning rows
print("Total records after removing duplicates:", len(df))

# Verify duplicates are removed
print("Remaining duplicates:", df.duplicated().sum())

```

# Step 3 — Convert Date and Time Columns

Now we'll convert StartDatetime and EndDatetime into proper Python datetime values.

```python

# Convert datetime columns
df["StartDatetime"] = pd.to_datetime(
    df["StartDatetime"]
)

df["EndDatetime"] = pd.to_datetime(
    df["EndDatetime"]
)

# Check converted data types
print(df[["StartDatetime", "EndDatetime"]].dtypes)

# Preview converted values
display(df[["StartDatetime", "EndDatetime"]].head())
```

# Step 4 Check Invalid Charging Durations


Check for missing, zero, and negative charging durations before making corrections.

```python

# Count negative durations
negative_duration = (df["Duration"] < 0).sum()

# Count zero durations
zero_duration = (df["Duration"] == 0).sum()

# Count positive durations
positive_duration = (df["Duration"] > 0).sum()

print("Negative durations:", negative_duration)
print("Zero durations:", zero_duration)
print("Positive durations:", positive_duration)
```

```python
df.duplicated().sum()
```

 # Step 5 Investigate Invalid Charging Durations

```python

# Display records with negative durations
df[df["Duration"] < 0].head(17)
```

```python
df[df["Duration"] == 0].head(20)
```

#  step 6 Validate Charging Duration Using Timestamps
Calculate duration from timestamps

```python
# Calculate charging duration in minutes
df["CalculateDuration"]=(
    df["EndDatetime"] - df["StartDatetime"]
    ).dt.total_seconds()/60     #dt.total_seconds() Converts the difference into seconds.

# Display zero-duration records for comparison
df[df["Duration"] == 0][
    ["StartDatetime","EndDatetime","Duration","CalculateDuration","Demand"]
    ].head(10)
```

```python

# Count zero-duration records with positive calculated duration
incorrect_zero = (
    (df["Duration"] == 0) &
    (df["CalculateDuration"] > 0)
).sum()

print("Zero-duration records with positive calculated duration:", incorrect_zero)
```

# Classify Zero-Duration Records

Objective: Identify how many zero-duration charging sessions have positive, zero, or negative calculated durations.

```python

# Select zero-duration records
zero_records = df[df["Duration"] == 0]

# Count each category
positive = (zero_records["CalculateDuration"] > 0).sum()
zero = (zero_records["CalculateDuration"] == 0).sum()
negative = (zero_records["CalculateDuration"] < 0).sum()

print("Positive calculated duration:", positive)
print("Zero calculated duration:", zero)
print("Negative calculated duration:", negative)
print("Total:", positive + zero + negative)
```

# step 7 Calculate Precise Charging Duration

Use the original date and time columns to calculate charging duration with second-level precision.

```python

# Combine date and time columns
start_full = pd.to_datetime(
    df["StartDay"] + " " + df["StartTime"]
)

end_full = pd.to_datetime(
    df["EndDay"] + " " + df["EndTime"]
)

# Calculate precise charging duration in minutes
df["PreciseDuration"] = (
    end_full - start_full
).dt.total_seconds() / 60

# Display zero-duration records
df[df["Duration"] == 0][
    ["StartTime", "EndTime", "Duration", "CalculateDuration", "PreciseDuration"]
].head(10)
```

# step 8 Identify Recoverable Zero Durations

Identify zero-duration records that have a positive duration calculated from precise timestamps.

```python

# Select zero-duration records
zero_records = df[df["Duration"] == 0]

# Classify using precise duration
positive = (zero_records["PreciseDuration"] > 0).sum()
zero = (zero_records["PreciseDuration"] == 0).sum()
negative = (zero_records["PreciseDuration"] < 0).sum()

print("Positive precise duration:", positive)
print("Zero precise duration:", zero)
print("Negative precise duration:", negative)
print("Total:", positive + zero + negative)
```

# Flag Negative Charging Durations

Identify records with negative charging durations so they can be handled during data cleaning.

```python
# Flag records with negative charging durations
df["InvalidNegativeDuration"] = (
    (df["Duration"] < 0) |
    (df["PreciseDuration"] < 0)
)

# Count flagged records
print("Invalid negative-duration records:",
      df["InvalidNegativeDuration"].sum())
```

# Flag Zero-Duration Charging Records

 Identify records with zero recorded duration and distinguish those with positive elapsed time from those with no elapsed time.

```python
# Flag zero-duration records that have positive elapsed time
df["ZeroDurationRecoverable"] = (
    (df["Duration"] == 0) &
    (df["PreciseDuration"] > 0)
)

# Flag records with both durations equal to zero
df["ZeroDurationUnresolved"] = (
    (df["Duration"] == 0) &
    (df["PreciseDuration"] == 0)
)

print("Recoverable records:", df["ZeroDurationRecoverable"].sum())
print("Unresolved records:", df["ZeroDurationUnresolved"].sum())
```

## Step 9 — Correct Zero-Duration Records

Correct recoverable zero durations using precise timestamps.
Mark unresolved and negative durations as missing.
Preserve the original Duration column.

```python

import numpy as np

# Create a copy of the original duration
df["CleanedDuration"] = df["Duration"].astype(float)

# Identify recoverable zero durations
recoverable_mask = (
    (df["Duration"] == 0) &
    (df["PreciseDuration"] > 0)
)

# Correct recoverable zero durations
df.loc[recoverable_mask, "CleanedDuration"] = (
    df.loc[recoverable_mask, "PreciseDuration"]
)

# Mark invalid and unresolved durations as missing
invalid_mask = (
    (df["Duration"] < 0) |
    (df["PreciseDuration"] < 0) |
    ((df["Duration"] == 0) & (df["PreciseDuration"] == 0))
)

df.loc[invalid_mask, "CleanedDuration"] = np.nan

# Verify cleaning results
print("Corrected zero durations:", recoverable_mask.sum())
print("Missing cleaned durations:", df["CleanedDuration"].isna().sum())
print("Negative cleaned durations:", (df["CleanedDuration"] < 0).sum())
```

# Step 10 — Validate Cleaned Duration

Verify that corrected durations are positive and invalid durations are marked as missing.

```python

# Validate cleaned duration values

print("Positive cleaned durations:",
      (df["CleanedDuration"] > 0).sum())

print("Zero cleaned durations:",
      (df["CleanedDuration"] == 0).sum())

print("Negative cleaned durations:",
      (df["CleanedDuration"] < 0).sum())

print("Missing cleaned durations:",
      df["CleanedDuration"].isna().sum())

print("Total records:", len(df))
```

# Step 11 — Validate Charging Demand

Check for missing, zero, and negative charging demand values.

```python

# Validate charging demand values

print("Missing demand values:",
      df["Demand"].isna().sum())

print("Zero demand values:",
      (df["Demand"] == 0).sum())

print("Negative demand values:",
      (df["Demand"] < 0).sum())

print("Positive demand values:",
      (df["Demand"] > 0).sum())

# Display demand statistics
print("\nDemand statistics:")
print(df["Demand"].describe())
```

# Step 12 — Final Validation

Verify the number of records, duplicates, missing values, and cleaned charging durations.

```python

# Final validation of cleaned dataset

print("Total records:", len(df))

print("Duplicate records:",
      df.duplicated().sum())

print("Missing cleaned durations:",
      df["CleanedDuration"].isna().sum())

print("Zero cleaned durations:",
      (df["CleanedDuration"] == 0).sum())

print("Negative cleaned durations:",
      (df["CleanedDuration"] < 0).sum())

print("Missing demand values:",
      df["Demand"].isna().sum())

print("Total columns:", len(df.columns))
```

# Step 13 — Save Cleaned Dataset

```python
import os
os.makedirs("Data_row/cleaned data", exist_ok=True)

df.to_csv("Data_row/cleaned data/ChargingRecords_Cleaned.csv", index=False)
```
