# relationship between Duration and Demand
Question: Do longer charging sessions generally have higher demand?

```python

# Calculate correlation
correlation = df[
    ["Duration", "Demand"]
].corr()

print(correlation)



```

```python

# Convert timestamps to datetime
start = pd.to_datetime(df["StartDatetime"])
end = pd.to_datetime(df["EndDatetime"])

# Calculate duration in minutes
calculated_duration = (
    end - start
).dt.total_seconds() / 60

# Create a comparison table
comparison = df[
    ["StartDatetime", "EndDatetime", "Duration", "Demand"]
].copy()

comparison["CalculatedDuration"] = calculated_duration

comparison["Difference"] = (
    comparison["Duration"] - comparison["CalculatedDuration"]
)

# Show first 10 zero-duration records
comparison[df["Duration"] == 0].head(10)
```
